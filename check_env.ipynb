# بررسی محیط اجرای پروژه (check_env)
این نوت‌بوک باید روی رایانه **خود دانشجو** اجرا شود تا نسخه واقعی پایتون و کتابخانه‌های نصب‌شده ثبت شود. خروجی‌ها عمداً از قبل پر نشده‌اند.

```python
import sys
import platform
import importlib
from importlib.metadata import version, PackageNotFoundError

print("Python version:", sys.version)
print("Python executable:", sys.executable)
print("Operating system:", platform.platform())

for package in ["numpy", "matplotlib", "pandas", "notebook", "ipykernel"]:
    try:
        print(f"{package}: {version(package)}")
    except PackageNotFoundError:
        print(f"{package}: not installed")
```

پس از اجرای همه سلول‌ها، نوت‌بوک را ذخیره کنید تا خروجی نسخه‌ها در GitHub ثبت شود. اگر کتابخانه‌ای نصب نیست، محیط مجازی یا conda خود را بررسی کنید.
